```python
# Получение данных из словаря с помощью методов

user = {
    "name": "Bob",
    "age": 32,
    "prof": "IT",
    "city": "Moscow",
}

keys = user.keys()
print(keys)

for key in keys:
    print(key)
```

```python
values = user.values()
print(values)

for value in values:
    print(value)
```

```python
items = user.items()
print(items)

for item in items:
    print(item)
```

```python
for key, value in user.items():
    print(key, value)
```

```python
# Динамическое обновление
user = {
    "name": "Bob",
    "age": 32,
    "prof": "IT",
    "city": "Moscow",
}

keys = user.keys()
print(keys)

user["last_name"] = "Ivanov"
print(keys)
```
