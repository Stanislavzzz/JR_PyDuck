# List comprehensions

```python
result = []
for i in range(1, 6):
    result.append(i * 2)
print(result)
```

```python
result = [i * 2 for i in range(1, 6)]
print(result)
```

```python
word = "Python"
list_word = []
for w in word:
    list_word.append(w)
print(list_word)
```

```python
list_word = [w for w in word]
print(list_word)
```

# Условное включение


```python
evens = []
for x in range(1, 21):
    if x % 2 == 0:
        evens.append(x)
print(evens)
```

```python
evens = [x for x in range(1, 21) if x % 2 == 0]
print(evens)
```

```python
words = ["word", "", "python", "", "", "hello"]
non_empty = []
for w in words:
    if w:  # if w != ""    или w == True
        non_empty.append(w)
print(non_empty)
```

```python
words = ["word", "", "python", "", "", "hello"]
non_empty = [w for w in words if w]
print(non_empty)
```

# Вложенные включения

```python
list_a = [1, 2, 3]
list_b = [10, 20, 30]
result = []
for x in list_a:
    for y in list_b:
        result.append((x, y))
print(result)
```

```python
list_a = [1, 2, 3]
list_b = [10, 20, 30]

result = [(x, y) for x in list_a for y in list_b]

print(result)
```

# Вложенные включения + фильтрация

```python
list_a = [1, 2, 3]
list_b = [10, 20, 30]
result = []
for x in list_a:
    for y in list_b:
        if x % 2 != 0:
            result.append((x, y))
print(result)
```

```python
list_a = [1, 2, 3]
list_b = [10, 20, 30]

result = [(x, y) for x in list_a for y in list_b if x % 2 != 0]

print(result)
```

# Тернарный оператор

A if условие else B

```python
x = 3
if x % 2 == 0:
    result = "Четное"
else:
    result = "Нечетное"
print(result)
```

```python
x = 4
result = "Четное" if x % 2 == 0 else "Нечетное"
print(result)
```

```python
nums = [1, 3, 4, 7, 2, 5, 9, 2, 3]
result = []
for n in nums:
    if n % 2 == 0:
        result.append(n)
    else:
        result.append(-n)
print(result)
```

```python
nums = [1, 3, 4, 7, 2, 5, 9, 2, 3]

result = [n if n % 2 == 0 else -n for n in nums]

print(result)
```

```python
nums = [1, 3, 4, 7, 2, 5, 9, 2, 3]
result = []
for n in nums:
    if n % 3 != 0:
        if n % 2 == 0:
            result.append(n)
        else:
            result.append(-n)
print(result)
```

```python
nums = [1, 3, 4, 7, 2, 5, 9, 2, 3]

result = [n if n % 2 == 0 else -n for n in nums if n % 3 != 0]

print(result)
```

# Set Comprehensions

```python
nums = [1, 3, 4, 7, 2, 5, 9, 2, 3]
result = set()
for n in nums:
    result.add(n)

print(result)
```

```python
nums = [1, 3, 4, 7, 2, 5, 9, 2, 3]
result = {n for n in nums}
print(result)
```

```python
nums = [1, 3, 4, 7, 2, 5, 9, 2, 3]
result = set()
for n in nums:
    if n % 2 != 0:
        result.add(n)

print(result)
```

```python
nums = [1, 3, 4, 7, 2, 5, 9, 2, 3]
result = {n for n in nums if n % 2 != 0}
print(result)
```

```python
result = set()
for num in range(1, 21):
    result.add(num % 3)
print(result)
```

```python
result = {num % 3 for num in range(1, 21)}
print(result)
```

```python
text = "Python PYTHON pyTHon 123123"
letters = set()
for char in text:
    if char.isalpha():
        letters.add(char.lower())
print(letters)
```

```python
text = "Python PYTHON pyTHon 123123"
letters = {char.lower() for char in text if char.isalpha()}
print(letters)
```

# Dict Comprehensions

```python
nums = [1, 3, 4, 7, 2, 5, 9, 2, 3]
result = {}
for n in nums:
    if n % 2 != 0:
        result[n] = n ** 2

print(result)
```

```python
nums = [1, 3, 4, 7, 2, 5, 9, 2, 3]
result = {n: n**2 for n in nums if n % 2 != 0}
print(result)
```

# Генераторные выражения


```python
nums = [1, 3, 4, 7]
result = (n**2 for n in nums if n % 2 != 0)
print(result)

print(next(result))
print(next(result))
print(next(result))
```

```python
nums = [1, 3, 4, 7]
result = (n**2 for n in nums if n % 2 != 0)
print(result)

print(list(result))
```

```python
nums = [1, 3, 4, 7]
result = (n**2 for n in nums if n % 2 != 0)
print(result)
for n in result:
    print(n)
```
