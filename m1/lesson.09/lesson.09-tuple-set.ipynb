```python
set_a = {1, 2, 3, 4, 5}
set_b = {3, 4, 5, 7, 9}
print(set_a)
print(type(set_a))
print(set_b)
print(type(set_b))
```

```python
# объединение
union_set = set_a.union(set_b)
print(union_set)
```

```python
union_set = set_a | set_b
print(union_set)
```

```python
# Пересечение
intersection_set = set_a.intersection(set_b)
print(intersection_set)

intersection_set = set_a & set_b
print(intersection_set)
```

```python
set_a = {1, 2, 3, 4, 5}
set_b = {3, 4, 5, 7, 9}
```

```python
# Разность
diff_set = set_a.difference(set_b)
print(diff_set)

diff_set = set_a - set_b
print(diff_set)
```

```python
# Разность
diff_set = set_b.difference(set_a)
print(diff_set)

diff_set = set_b - set_a
print(diff_set)
```

```python
# Симметрическая разность
diff_set = set_a.symmetric_difference(set_b)
print(diff_set)

diff_set = set_a ^ set_b
print(diff_set)
```
