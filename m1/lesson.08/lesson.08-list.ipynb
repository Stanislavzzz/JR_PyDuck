```python
names = ['Bob', 'Anna', 'Mary', 'Ivan', 'Sonya', [1, 2, 3, 4, 5]]
new_list = names.copy()

new_list[-1][1] = "Anton"
print(new_list)
print(id(new_list))
print(names)
print(id(names))
```

```python
from copy import deepcopy

names = ['Bob', 'Anna', 'Mary', 'Ivan', 'Sonya', [1, 2, 3, 4, 5]]
new_list = deepcopy(names)

new_list[-1][1] = "Anton"
print(new_list)
print(id(new_list))
print(names)
print(id(names))
```
