```python
age = int(input("Введите ваш возраст: "))

if age > 0 and age <= 7:
    print("Детский сад")
else:
    if age > 7 and age <= 18:
        print("Школа")
    else:
        if age > 18 and age <= 25:
            print("Студент")
        else:
            if age > 25 and age <= 60:
                print("Взрослый")
            else:
                if age > 60:
                    print("Пенсионер")
                else:
                    print("Не верный возраст")
```
