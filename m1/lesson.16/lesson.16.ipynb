```python
from math import pi


class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return pi * self.radius ** 2

    def perimeter(self):
        return 2 * pi * self.radius


class Square:
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side

    def perimeter(self):
        return 4 * self.side
```

```python
def prin_shape_info(shape):
    print(f"Area={shape.area()}, Perimeter={shape.perimeter()}")
```

```python
square = Square(10)
prin_shape_info(square)

circle = Circle(10)
prin_shape_info(circle)
```

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

```

```python
rect = Rectangle(10, 15)
prin_shape_info(rect)
```

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def info(self):
        print(f"Vehicle brand: {self.brand}")


class Car(Vehicle):
    def __init__(self, brand, seats):
        super().__init__(brand)
        self.seats = seats

    def open_trunk(self):
        print("Trunk opened")


car = Car("Toyota", seats=5)
print(car.brand, car.seats)
car.info()
car.open_trunk()
```

```python
from math import pi
from abc import ABC, abstractmethod


class Shape(ABC):
    """Абстрактный класс фигуры."""

    @abstractmethod
    def area(self):
        """Абстрактный класс площади."""
        pass

    @abstractmethod
    def perimeter(self):
        """Абстрактный класс периметр."""
        pass


class Circle(Shape):
    """Касс круг."""

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        """Касс круг."""
        return pi * self.radius ** 2

    def perimeter(self):
        """Касс круг."""
        return 2 * pi * self.radius


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side

    def perimeter(self):
        return 4 * self.side
```

```python
class LoggerMixin:
    def log(self, message):
        print(f"LOG: {message}")

    def info(self):
        print("Logger!!!")


class Page:
    def __init__(self, content):
        self.content = content

    # def info(self):
    #     print("Page!!!")


# Этот класс наследует и от Page, и от LoggerMixin
class LoggablePage(Page, LoggerMixin):
    def show_and_log(self):
        self.log(f"Showing page content: {self.content}")
        print(self.content)

    # def info(self):
    #     print("my info")


page = LoggablePage("My secret content")
page.show_and_log()
page.info()
```
