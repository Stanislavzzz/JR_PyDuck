```python
class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def get_name(self):
        return self.__name

    def set_name(self, name1):
        self.__name = name1

    def get_age(self):
        return self.__age

    def set_age(self, age1):
        if age1 > 0:
            self.__age = age1
        else:
            self.__age = 0

p = Person('Bob', 32)

print(p.get_name(), p.get_age())
p.set_name("Anna")
p.set_age(53)
print(p.get_name(), p.get_age())
```

```python
class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    @property
    def name(self):
        return self.__name

    @name.setter
    def name(self, name1):
        if name1 != "Admin":
            self.__name = name1
        else:
            self.__name = "User"

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, age1):
        if age1 > 0:
            self.__age = age1
        else:
            self.__age = 0

p = Person('Bob', 32)

print(p.name, p.age)
p.name = "Admin"
p.age = -10
print(p.name, p.age)
```

```python
from math import pi


class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return pi * self.radius ** 2

    def perimeter(self):
        return 2 * pi * self.radius


class Square:
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side

    def perimeter(self):
        return 4 * self.side
```
